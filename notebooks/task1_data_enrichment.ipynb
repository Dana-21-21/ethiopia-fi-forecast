```python
import pandas as pd
import numpy as np

file_path = "../data/raw/ethiopia_fi_unified_data.xlsx"

df = pd.read_excel(
    file_path,
    sheet_name="ethiopia_fi_unified_data"
)

impact_df = pd.read_excel(
    file_path,
    sheet_name="Impact_sheet"
)

print(df.shape)
print(impact_df.shape)
```

```python
new_observations = pd.DataFrame([
    {
        "record_id": "REC_NEW_001",
        "record_type": "observation",
        "category": None,
        "pillar": "USAGE",
        "indicator": "Telebirr Registered Users",
        "indicator_code": "USG_TELEBIRR_USERS",
        "indicator_direction": "higher_better",
        "value_numeric": 54,
        "value_text": "54 million users",
        "value_type": "million_users",
        "unit": "million users",
        "observation_date": pd.to_datetime("2025-01-01"),
        "gender": "all",
        "location": "national",
        "source_name": "Ethio Telecom",
        "source_type": "operator",
        "confidence": "high",
        "source_url": "",
        "original_text": "Telebirr reached over 54 million users",
        "collected_by": "Your_Name",
        "collection_date": pd.to_datetime("2026-07-19"),
        "notes": "Added as a leading indicator for digital payment adoption forecasting"
    }
])
```

```python
new_observations
```

```python
df_enriched = pd.concat(
    [
        df,
        new_observations
    ],
    ignore_index=True
)
```

```python
df_enriched.shape
```

```python
import os

os.makedirs(
    "../data/processed",
    exist_ok=True
)
```

```python
output_path = "../data/processed/ethiopia_fi_enriched_data.xlsx"

df_enriched.to_excel(
    output_path,
    index=False
)
```
